```bash
import pandas as pd
import ipywidgets as widgets
from IPython.display import display
import json

# 预设变量和路径
pulsar_cluster_name = "pulsar_cluster_1"
sub_dir = "broker1"
config_file_path = f"{pulsar_cluster_name}/{sub_dir}/conf/broker.conf"

# 读取配置文件
def read_config(file_path):
    config_dict = {}
    try:
        with open(file_path, 'r') as file:
            lines = file.readlines()
            for line in lines:
                if "=" in line and not line.startswith('#'):
                    key, value = line.strip().split('=', 1)
                    config_dict[key] = value
    except Exception as e:
        print(f"Error reading the config file: {e}")
    return config_dict

# 保存配置文件
def save_config(config_dict, file_path):
    try:
        with open(file_path, 'w') as file:
            for key, value in config_dict.items():
                file.write(f"{key}={value}\n")
        print("Configuration saved successfully!")
    except Exception as e:
        print(f"Error saving the config file: {e}")

# 创建可编辑的配置表
def create_editable_config_table(config_dict):
    items = [widgets.Text(value=v, description=k) for k, v in config_dict.items()]
    return items

# 显示配置
def display_config():
    config_dict = read_config(config_file_path)
    if config_dict:
        items = create_editable_config_table(config_dict)
        save_button = widgets.Button(description="Save Changes")
        save_button.on_click(lambda b: save_config({item.description: item.value for item in items}, config_file_path))
        display(*items, save_button)
    else:
        print("No configuration data found or error in reading the file.")

# 创建和显示按钮
button = widgets.Button(description="Load and Edit Config")
output = widgets.Output()

def on_button_clicked(b):
    with output:
        output.clear_output()
        display_config()

button.on_click(on_button_clicked)
display(button, output)
```
